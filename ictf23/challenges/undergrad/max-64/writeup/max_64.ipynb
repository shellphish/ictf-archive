```python
from PIL import Image, ImageOps
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
```

```python
img = Image.open('max_64.jpeg')
plt.imshow(img)
```

```python
import tensorflow as tf

model = tf.keras.Sequential([
    tf.keras.layers.MaxPool2D(pool_size=64, strides=64)
])
```

```python
img_arr = np.array(img).reshape(1, 224, 224, 3)
```

```python
output = model.predict(img_arr).reshape(3,3,3)
```

```python
flag = output.reshape(1, 27)
print(flag)
```
